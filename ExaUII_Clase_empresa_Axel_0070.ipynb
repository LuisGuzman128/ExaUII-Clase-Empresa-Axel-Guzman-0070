```python
# ==========================================
# Actividad: Clase Empresa
# Alumno: Axel Guzman
# Matrícula: 0070
# ==========================================

class Empresa:
    def __init__(self):
        # Atributos solicitados
        self.razonSocial = ""
        self.sector = ""
        self.numEmpleados = 0

    def leerDatos(self):
        """Función para capturar datos desde el teclado"""
        print(f"\n--- Captura de datos de la Empresa ---")
        self.razonSocial = input("Ingrese la Razón Social: ")
        self.sector = input("Ingrese el Sector Económico: ")
        self.numEmpleados = int(input("Ingrese el número de empleados: "))

    def mostrarDatos(self):
        """Función para imprimir los datos del objeto"""
        print("\n" + "="*30)
        print(f"FICHA TÉCNICA: {self.razonSocial}")
        print("="*30)
        print(f"Sector:      {self.sector}")
        print(f"Empleados:   {self.numEmpleados}")
        print("="*30)

# --- Creación de los 2 Objetos solicitados ---

# Instancia del primer objeto
empresa1 = Empresa()
print("Configurando Empresa 1:")
empresa1.leerDatos()

# Instancia del segundo objeto
empresa2 = Empresa()
print("\nConfigurando Empresa 2:")
empresa2.leerDatos()

# --- Ejecución del programa interactivo ---
print("\n" + "#"*40)
print("REPORTE FINAL DE OBJETOS CREADOS")
print("#"*40)

empresa1.mostrarDatos()
empresa2.mostrarDatos()
```
